```python
def canonical_employment_status(value):

    if value in EMPLOYED_Q30:
        return "employed"

    if value in ACTIVE_UNEMPLOYED_Q30:
        return "active_unemployed"

    if value in NOT_LOOKING_Q30:
        return "not_looking"

    if value in STUDENT_Q30:
        return "full_time_student"

    return np.nan


df["employment_status_canonical"] = (
    df["employment_status"]
    .apply(canonical_employment_status)
)

print(
    df["employment_status_canonical"]
    .value_counts(dropna=False)
)
```

```python
YES_VALUES = {
    "Yes",
    "ඔව්",
    "ஆம்"
}

NO_VALUES = {
    "No",
    "නැත",
    "இல்லை"
}


def canonical_yes_no(value):

    if value in YES_VALUES:
        return "yes"

    if value in NO_VALUES:
        return "no"

    return np.nan


df["recent_job_search_action_canonical"] = (
    df["recent_job_search_action"]
    .apply(canonical_yes_no)
)

df["available_within_two_weeks_canonical"] = (
    df["available_within_two_weeks"]
    .apply(canonical_yes_no)
)
```

```python
def assign_labour_market_group(row):

    status = row["employment_status_canonical"]
    recent_search = row["recent_job_search_action_canonical"]
    available = row["available_within_two_weeks_canonical"]

    # Class 1 candidate
    if status == "employed":
        return "EMPLOYED"

    # Strict primary negative class:
    # actively unemployed + evidence of recent search activity
    if (
        status == "active_unemployed"
        and recent_search == "yes"
    ):
        return "ACTIVE_UNEMPLOYED_RECENT_SEARCH"

    # Actively-unemployed self-report but no recent search action.
    # Keep separate for sensitivity analysis.
    if (
        status == "active_unemployed"
        and recent_search == "no"
        and available == "yes"
    ):
        return "AVAILABLE_NOT_RECENTLY_SEARCHING"

    if (
        status == "active_unemployed"
        and recent_search == "no"
        and available == "no"
    ):
        return "NOT_SEARCHING_NOT_AVAILABLE"

    if status == "not_looking":
        return "OUTSIDE_PRIMARY_LABOUR_FORCE"

    if status == "full_time_student":
        return "FULL_TIME_STUDENT"

    return "UNCLASSIFIED"


df["labour_market_group"] = df.apply(
    assign_labour_market_group,
    axis=1
)

print(
    df["labour_market_group"]
    .value_counts()
)
```

```python
PRIMARY_GROUPS = {
    "EMPLOYED",
    "ACTIVE_UNEMPLOYED_RECENT_SEARCH"
}

df["eligible_primary_model"] = (
    df["labour_market_group"]
    .isin(PRIMARY_GROUPS)
)

df["employment_outcome"] = np.where(
    df["labour_market_group"] == "EMPLOYED",
    1,
    np.where(
        df["labour_market_group"] == "ACTIVE_UNEMPLOYED_RECENT_SEARCH",
        0,
        np.nan
    )
)

print("Current-schema respondents:", len(df))

print(
    "\nLabour-market groups:"
)

print(
    df["labour_market_group"]
    .value_counts()
)

print(
    "\nEligible for primary binary model:",
    df["eligible_primary_model"].sum()
)
```

```python
def extract_likert_number(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if len(value) > 0 and value[0] in {"1", "2", "3", "4", "5"}:
        return int(value[0])

    return np.nan


for col in [
    "english_communication",
    "digital_confidence",
]:
    df_model[col] = df_model[col].apply(
        extract_likert_number
    )

for col in [
    "english_communication",
    "digital_confidence",
]:
    print("\n", col)
    print(df_model[col].value_counts(dropna=False).sort_index())
```
